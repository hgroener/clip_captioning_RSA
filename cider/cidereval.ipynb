```python
# demo script for running CIDEr
from pydataformat.loadData import LoadData
import pdb
import json
from pyciderevalcap.eval import CIDErEvalCap as ciderEval
import pandas as pd
import os

#pathToData = 'data/'
pathToData = "/home/user/Documents/Projekt_Sprachgenerierung/project/clip_captioning_RSA/data/AbstractScenes_v1.1/debug/"

#refName = 'pascal50S.json'
refName = "gold_caps.json"
#candName = 'pascal_candsB.json'
candName = "gen_caps.json"

result_file = 'results.json'
df_mode = 'corpus'
```

```python
df = pd.read_csv("/home/user/Documents/Projekt_Sprachgenerierung/project/clip_captioning_RSA/data/AbstractScenes_v1.1/df_debug.csv")
```

```python
gold_caps = []
gen_caps = []

for i, row in df.iterrows():
    gold = [{"image_id": row["file"], "caption": row["cap" + str(i)]} for i in range(6)]
    gen = {"image_id": row["file"], "caption": row["cap_RSA_t"]}
    
    gold_caps += gold
    gen_caps.append(gen)
    
os.makedirs("/home/user/Documents/Projekt_Sprachgenerierung/project/clip_captioning_RSA/data/AbstractScenes_v1.1/debug/")
    
with open("/home/user/Documents/Projekt_Sprachgenerierung/project/clip_captioning_RSA/data/AbstractScenes_v1.1/debug/gold_caps.json", "w") as f:
    json.dump(gold_caps, f)
with open("/home/user/Documents/Projekt_Sprachgenerierung/project/clip_captioning_RSA/data/AbstractScenes_v1.1/debug/gen_caps.json", "w") as f:
    json.dump(gen_caps, f)
```

```python
# load reference and candidate sentences
loadDat = LoadData(pathToData)
gts, res = loadDat.readJson(refName, candName)
```

```python
# calculate cider scores
#gts = ["testing", "this", "with", "random", "sents"]
#distr1 = ["I", "am"] + gts
#distr2 = gts + ["is", "it", "working"]

#res = [distr1, distr2]
scorer = ciderEval(gts, res, df_mode)
# scores: dict of list with key = metric and value = score given to each candidate
#scores = scorer.evaluate()
```

```python
scores = scorer.evaluate()
```

```python
# scores['CIDEr'] contains CIDEr scores
# scores['CIDErD'] contains CIDEr-D scores

with open(result_file, 'w') as outfile:
    json.dump(scores, outfile)
```
